```python
def predict_next_day(past_temperatures, window=3):
    recent = past_temperatures[-window:]
    prediction = sum(recent)/len(recent)
    return round(prediction, 1)
if __name__ == "__main__":
    past_week = [30,100,29,7,33,3,30]
    actual_tomorrow = 31
    predicted = predict_next_day(past_week, window=3)
    error = round(abs(predicted - actual_tomorrow), 1)
    print (f"Predicted: {predicted}C")
    print (f"Actual: {actual_tomorrow}C")
    print (f"Error: {error}C")
```

```python
def weather_advice(temperature_c, humidity_percent):
    """Rule-based weather advisor using simple IF/THEN logic."""
    advice = []

    if temperature_c >= 35:
        advice.append("Extreme heat: stay hydrated and avoid sun.")
    if temperature_c <=5:
        advice.append("Cold weather: wear a warm jacket.")
    if temperature_c >= 40:
        advice.append("Very extreme heat: stay indoors if possible.")
    if temperature_c >= 30 and humidity_percent >= 70:
        advice.append("Hot and humid: avoid heavy physical activity.")
    if temperature_c <-10 or temperature_c >= 50:
        advice.append("Too Much Temperature.")
    if temperature_c <= 0:
        advice.append("Freezing temperature: protect yourself from frost.")
    if temperature_c <= 10 and humidity_percent >= 70:
        advice.append("Cold and humid: wear warm and water-resistant clothing.")
    if humidity_percent <= 30:
        advice.append("Low humidity: drink water and use moisturizer.")
    if humidity_percent >=80:
        advice.append("High humidity: expect possible rain, carry an umbrella.")
    if not advice:
        advice.append("Conditions look moderate. No special precautions needed.")
    return advice

if __name__ == "__main__":
    sample_days = [(28, 25),(32, 20),(27, 30),(35, 25),(30, 28)]
    for temp, humidity in sample_days:
        result = weather_advice(temp, humidity)
        print(f"Temp={temp}C, Humidity={humidity}% -> {result}")
```

```python
def classify_email(email):
    spam_keywords = [
        "free",
        "winner",
        "urgent",
        "click here",
        "prize",
        "offer",
        "congratulations",
        "money"
    ]

    email = email.lower()

    count = 0

    for keyword in spam_keywords:
        if keyword in email:
            count += 1

    if count >= 3:
        classification = "Spam"
    else:
        classification = "Legitimate"

    return classification, count


if __name__ == "__main__":

    sample_emails = [
        (
            "Congratulations! You are a winner. Click here to claim your free prize!",
            "Spam"
        ),
        (
            "Urgent! You have won a free prize. Click here for your money.",
            "Spam"
        ),
        (
            "Special offer! Congratulations, you are a winner. Claim your free prize.",
            "Spam"
        ),
        (
            "Hi, can you send me the notes from today's lecture?",
            "Legitimate"
        ),
        (
            "Your project meeting is scheduled for tomorrow at 10 AM.",
            "Legitimate"
        ),
        (
            "Please review the assignment and submit it before Friday.",
            "Legitimate"
        )
    ]

    correct_predictions = 0

    print("===== Spam Classifier Results =====\n")

    for i, (email, actual_label) in enumerate(sample_emails, 1):

        predicted_label, keyword_count = classify_email(email)

        if predicted_label == actual_label:
            correct_predictions += 1

        print(f"Test {i}:")
        print(f"Email: {email}")
        print(f"Spam Keywords Found: {keyword_count}")
        print(f"Predicted: {predicted_label}")
        print(f"Actual: {actual_label}")
        print("-" * 50)

    accuracy = (correct_predictions / len(sample_emails)) * 100

    print(f"\nCorrect Predictions: {correct_predictions}/{len(sample_emails)}")
    print(f"Accuracy: {accuracy:.2f}%")
```

```python
def classify_email(email, threshold=3):
    spam_keywords = [
        "free",
        "winner",
        "urgent",
        "click here",
        "prize",
        "offer",
        "congratulations",
        "money"
    ]

    email = email.lower()

    count = 0

    for keyword in spam_keywords:
        if keyword in email:
            count += 1

    if count >= threshold:
        classification = "Spam"
    else:
        classification = "Legitimate"

    return classification, count


if __name__ == "__main__":

    sample_emails = [
        (
            "Congratulations! You are a winner. Click here to claim your free prize!",
            "Spam"
        ),
        (
            "Urgent! You have won a free prize. Click here for your money.",
            "Spam"
        ),
        (
            "Special offer! Congratulations, you are a winner. Claim your free prize.",
            "Spam"
        ),
        (
            "Hi, can you send me the notes from today's lecture?",
            "Legitimate"
        ),
        (
            "Your project meeting is scheduled for tomorrow at 10 AM.",
            "Legitimate"
        ),
        (
            "Please review the assignment and submit it before Friday.",
            "Legitimate"
        )
    ]


    print("========== TASK 2 ==========")

    threshold = 3
    correct_predictions = 0

    for i, (email, actual_label) in enumerate(sample_emails, 1):

        predicted_label, keyword_count = classify_email(email, threshold)

        if predicted_label == actual_label:
            correct_predictions += 1

        print(f"\nTest {i}")
        print(f"Email: {email}")
        print(f"Keywords Found: {keyword_count}")
        print(f"Predicted: {predicted_label}")
        print(f"Actual: {actual_label}")

    accuracy = (correct_predictions / len(sample_emails)) * 100

    print(f"\nCorrect Predictions: {correct_predictions}/{len(sample_emails)}")
    print(f"Accuracy: {accuracy:.2f}%")


    # ---------------- TASK 3 ----------------
    print("\n========== TASK 3 ==========")
    print("Testing different spam-keyword thresholds...\n")

    thresholds = [1, 2, 3, 4, 5, 6]

    for threshold in thresholds:

        correct_predictions = 0

        for email, actual_label in sample_emails:

            predicted_label, keyword_count = classify_email(
                email, threshold
            )

            if predicted_label == actual_label:
                correct_predictions += 1

        accuracy = (correct_predictions / len(sample_emails)) * 100

        print(
            f"Threshold = {threshold} "
            f"-> Accuracy = {accuracy:.2f}%"
        )
```
